## **몬테카를로 시뮬레이션**

## "무작위성과 확률의 무서움, 그리고 컴퓨터 연산의 위대함"을 동시에 보여주는 것
### **computational algorithms based on repeated random sampling for obtaining numerical results**

### ❶ 통계치의 요동 = "섣부른 일반화의 오류"를 경고
- 표본 데이터의 개수가 적을 경우는 우연히 어떤 값이 뽑히느냐에 따라 전체 집단의 성격이 왜곡
- 의미: 우리가 일상에서 "내가 해봤는데 그렇더라"라며 몇 번의 경험(적은 샘플)만으로 결론을 내리는 것이 얼마나 위험한 통계적 오류인지를 보여줌

### ❷ 대수의 법칙 (Law of Large Numbers)의 시각화
- 샘플을 8개 → 100개 → 1000개 → 8192개로 늘려가면, 상단의 날뛰던 숫자들(mu, sigma, skew, kurtosis)이 자석에 이끌리듯 점차 정규분포의 고정된 이론값으로 수렴
- 개별 난수는 어디로 튈지 모르는 '불확실성'을 가지지만, 그 난수가 거대하게 모이면 완벽한 '질서(정규분포)'를 만들어낸다는 대자연의 통계적 법칙을 시각적으로 증명하는 것

### ❸ 몬테카를로 방법의 신뢰성 확보
- 아래 애니메이션은 몬테카를로 알고리즘을 쓰는 작업자에게 "불안해하지 마라. 컴퓨터를 시켜서 무식하게 많이 돌리기만 하면(수만 번 샘플링), 수학 공식으로 힘들게 풀어낸 참값에 무조건 도달한다"는 확신과 신뢰성을 주는 시각적 장치

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.stats import norm
from IPython.display import HTML

# 1. 기본 설정
true_mu = 501.47
true_sigma = 91.71
sample_sizes = np.unique(np.logspace(0, 4, num=50, dtype=int)) # 프레임 수를 50개로 최적화

fig, ax = plt.subplots(figsize=(8, 5))

def update(frame):
    ax.clear()
    num_samples = sample_sizes[frame]

    np.random.seed(42)
    samples = np.random.normal(true_mu, true_sigma, num_samples)

    current_mu = np.mean(samples)
    current_sigma = np.std(samples)

    if num_samples > 2:
        from scipy.stats import skew, kurtosis
        current_skew = skew(samples)
        current_kurt = kurtosis(samples)
    else:
        current_skew, current_kurt = 0.0, 0.0

    # 이론적 정규분포 곡선
    x = np.linspace(100, 900, 1000)
    ax.plot(x, norm.pdf(x, true_mu, true_sigma), color='green', linewidth=1.5, label='normal')

    # 샘플 히스토그램
    bins_count = min(100, max(5, num_samples // 10))
    ax.hist(samples, bins=bins_count, density=True, histtype='step', color='red', alpha=0.8, label=f'{num_samples} samples')

    ax.axvline(true_mu, color='dodgerblue', linestyle='-', linewidth=0.8, alpha=0.5)

    title_text = f"mu: {current_mu:.4f}, sigma: {current_sigma:.4f}\nskew: {current_skew:.4f}, kurtosis: {current_kurt:.4f}"
    ax.set_title(title_text, fontsize=10, fontfamily='monospace')

    ax.set_xlim(100, 900)
    ax.set_ylim(0, 0.015)
    ax.legend(loc='upper right')

# 애니메이션 객체 생성
ani = FuncAnimation(fig, update, frames=len(sample_sizes), interval=150, repeat=False)

# 주피터 노트북 내에서 HTML5 비디오로 강제 변환하여 재생 (가장 확실함)
plt.close() # 불필요한 정적 창 닫기
HTML(ani.to_html5_video())
```
