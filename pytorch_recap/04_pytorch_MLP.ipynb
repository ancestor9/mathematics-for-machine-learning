```python
import torch
import torch.nn as nn

# MLP 모델 클래스 정의
class SimpleMLP(nn.Module):
    def __init__(self):
        super().__init__()
        # 첫 번째 은닉층: 10개의 입력 뉴런, 32개의 출력 뉴런
        self.hidden1 = nn.Linear(in_features=10, out_features=32)
        # 두 번째 은닉층: 32개의 입력 뉴런, 16개의 출력 뉴런
        self.hidden2 = nn.Linear(in_features=32, out_features=16)
        # 출력층: 16개의 입력 뉴런, 1개의 출력 뉴런
        self.output = nn.Linear(in_features=16, out_features=1)
        # 활성화 함수: ReLU (렐루)
        self.relu = nn.ReLU()

    def forward(self, x):
        # 순전파 과정
        x = self.relu(self.hidden1(x))
        x = self.relu(self.hidden2(x))
        x = self.output(x)
        return x

# 모델 객체 생성
model = SimpleMLP()
print(model)

# 임의의 입력 데이터 생성 (1개 샘플, 10개 특성)
random_input = torch.randn(1, 10)
# 모델에 입력 데이터 통과
output = model(random_input)
print(f"\n입력 데이터 크기: {random_input.shape}")
print(f"모델 출력 크기: {output.shape}")
```

## Perceptron with single input

```python
import torch
import torch.nn as nn

# 모델 클래스 정의
class Simple(nn.Module):
    def __init__(self):
        super().__init__()
        # 실제 정의: 1개의 입력 뉴런, 1개의 출력 뉴런
        self.hidden1 = nn.Linear(in_features=1, out_features=1)
        # 출력층: 1개의 입력 뉴런, 1개의 출력 뉴런
        self.output = nn.Linear(in_features=1, out_features=1)
        # 활성화 함수: ReLU (렐루)
        self.relu = nn.ReLU()

    def forward(self, x):
        # 순전파 과정
        x = self.relu(self.hidden1(x))
        x = self.output(x)
        return x

# 모델 객체 생성
perceptron = Simple()
print(perceptron)

# 임의의 입력 데이터 생성 (1개 샘플, 1개 특성) - perceptron 모델에 맞춤
random_input = torch.randn(1, 1)
# 모델에 입력 데이터 통과
# SimpleMLP 모델이 아닌, 새로 정의한 Simple 모델(perceptron)을 사용합니다.
output = perceptron(random_input)
print(f"\n입력 데이터 크기: {random_input.shape}")
print(f"모델 출력 크기: {output.shape}")
```

## Perceptron with multiple inputs

## 실전문제

- 아래 네트웤 구조를 갖는 인공신경망 모델을 class로 만들어라

<img src="https://cdn.prod.website-files.com/6641d3051a4e3c3f8f0e307d/68c77c0133c4aea224d71faf_image-1.png">

```python
import torch
import torch.nn as nn

class ImageBasedNet(nn.Module):
    def __init__(self):
        super(ImageBasedNet, self).__init__()
        # Layer 1: 입력층(3) -> 은닉층 1(4)
        self.fc1 = nn.Linear(3, 4)
        # Layer 2: 은닉층 1(4) -> 은닉층 2(4)
        self.fc2 = nn.Linear(4, 4)
        # Layer 3: 은닉층 2(4) -> 출력층(1)
        self.fc3 = nn.Linear(4, 1)
        # 활성화 함수 (ReLU 사용)
        self.relu = nn.ReLU()

    def forward(self, x):
        # --- 순전파(Forward Propagation) 과정 ---
        # 1. 입력데이터 -> fc1 -> ReLU
        x = self.relu(self.fc1(x))
        # 2. fc1의 결과 -> fc2 -> ReLU
        x = self.relu(self.fc2(x))
        # 3. fc2의 결과 -> fc3 (최종 출력)
        # 문제의 종류(회귀, 이진 분류 등)에 따라 활성화 함수(Sigmoid 등)가 추가될 수 있습니다.
        # 기본적인 선형 출력을 위해 그대로 둡니다.
        x = self.fc3(x)

        return x

# 3. 모델 인스턴스 생성 및 구조 확인
model = ImageBasedNet()
print(model)

# 4. (참고) 가상의 입력 데이터로 출력 확인해보기
# 입력 노드가 3개이므로 차원이 3인 데이터를 넣어야 합니다. (배치 크기는 1로 가정)
dummy_input = torch.randn(1, 3)
output = model(dummy_input)

print(f"\n입력 데이터: {dummy_input}")
print(f"모델 출력: {output}")
```
