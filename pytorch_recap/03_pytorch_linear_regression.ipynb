Step 1: 데이터 준비

먼저, y=2x+1이라는 실제 직선을 가정하고, 이 직선 주변에 약간의 노이즈(오차)를 더한 가상의 데이터를 만듭니다.

```python
import torch

# 실제 데이터 (ground truth)
w_true = 2.0
b_true = 1.0

# 100개의 임의의 x 값 생성
X = torch.randn(100) * 10

# y 값 생성: y = 2x + 1 + 노이즈
Y = w_true * X + b_true + torch.randn(100) * 2
```

Step 2: 모델 설계

PyTorch에서 모델을 정의할 때는 torch.nn.Module 클래스를 상속받아 사용합니다. 이 클래스는 모델의 파라미터와 구조를 관리하는 데 필요한 기본 기능을 제공합니다.



```python
import torch.nn as nn

class LinearRegression(nn.Module):
    def __init__(self):
        super().__init__()
        # 학습할 파라미터 w와 b를 정의합니다.
        # nn.Parameter는 텐서를 모델의 파라미터로 등록하는 특별한 클래스입니다.
        self.w = nn.Parameter(torch.randn(1, requires_grad=True))
        self.b = nn.Parameter(torch.randn(1, requires_grad=True))

    def forward(self, x):
        # 모델의 순전파(forward pass)를 정의합니다.
        # 입력 x에 대해 예측값 y_pred를 계산합니다.
        return self.w * x + self.b

# 모델 객체 생성
model = LinearRegression()
```

Step 3: 손실 함수 및 옵티마이저 정의

손실 함수(Loss Function): 모델의 예측값과 실제 정답 사이의 오차를 계산하는 함수입니다. 선형 회귀에서는 주로 평균 제곱 오차(Mean Squared Error, MSE)를 사용합니다. PyTorch에서는 nn.MSELoss로 제공됩니다.

옵티마이저(Optimizer): 손실을 최소화하기 위해 모델의 파라미터(w, b)를 업데이트하는 알고리즘입니다. 여기서는 확률적 경사 하강법(Stochastic Gradient Descent, SGD)을 사용합니다. PyTorch에서는 torch.optim.SGD로 제공됩니다.

```python
import torch.optim as optim

# 손실 함수 정의
loss_fn = nn.MSELoss()

# 옵티마이저 정의
# model.parameters()는 모델의 모든 학습 가능한 파라미터를 반환합니다.
# lr (learning rate): 한 번에 파라미터를 얼마나 크게 업데이트할지 결정하는 학습률입니다.
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

Step 4: 모델 학습

모델을 반복적으로 학습시키는 과정을 에포크(Epoch)라고 부릅니다. 한 번의 에포크는 전체 데이터를 한 바퀴 도는 것을 의미합니다.

```python
# 학습 루프
num_epochs = 1000

for epoch in range(num_epochs):
    # 1. 순전파 (Forward Pass)
    y_pred = model(X)

    # 2. 손실 계산
    loss = loss_fn(y_pred, Y)

    # 3. 역전파 (Backward Pass)
    # 옵티마이저에 저장된 이전 기울기를 초기화합니다.
    optimizer.zero_grad()
    # 손실에 대한 모든 파라미터의 미분값(기울기)을 계산합니다.
    loss.backward()

    # 4. 파라미터 업데이트
    # 계산된 기울기를 사용하여 파라미터를 업데이트합니다.
    optimizer.step()

    # 100 에포크마다 손실 출력
    if (epoch + 1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')
```

Step 5: 모델 평가

학습이 끝난 후, 모델의 최종 파라미터(w, b)와 실제 파라미터(w_true, b_true)를 비교하여 모델이 잘 학습되었는지 확인해 봅시다.

```python
# 학습된 모델의 최종 파라미터
print("\n--- 최종 파라미터 ---")
print(f"학습된 w: {model.w.item():.4f}")
print(f"학습된 b: {model.b.item():.4f}")
print("---------------------")
print(f"실제 w: {w_true}")
print(f"실제 b: {b_true}")
```

## 실습하기

- torch.randn(100) * 2의 값을 torch.randn(100) * 10으로 변경하여 노이즈의 크기를 키웠을 때, 모델의 학습 결과가 어떻게 달라지는지 확인해 보세요.
