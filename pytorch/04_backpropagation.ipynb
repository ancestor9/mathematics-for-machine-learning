```python
import torch

x = torch.tensor(1.0)
y = torch.tensor(2.0)

# This is the parameter we want to optimize -> requires_grad=True
w = torch.tensor(1.0, requires_grad=True)

# forward pass to compute loss
y_predicted = w * x
loss = (y_predicted - y)**2
print(loss)
```

```python
# backward pass to compute gradient dLoss/dw
loss.backward()
print(w.grad)

# update weights
# next forward and backward pass...

# continue optimizing:
# update weights, this operation should not be part of the computational graph
with torch.no_grad():
    w -= 0.01 * w.grad
# don't forget to zero the gradients
w.grad.zero_()

# next forward and backward pass...
```

```python
import sympy

# Define symbolic variables
x_sym, y_sym, w_sym = sympy.symbols('x y w')

# Define the symbolic expressions for y_predicted and loss
y_predicted_sym = w_sym * x_sym
loss_sym = (y_predicted_sym - y_sym)**2

# Calculate the partial derivative of loss with respect to w symbolically
dloss_dw_sym = sympy.diff(loss_sym, w_sym)

print(f"Symbolic loss function: {loss_sym}")
print(f"Symbolic partial derivative of loss with respect to w (dLoss/dw): {dloss_dw_sym}")

# Verify with numerical values from the previous torch example:
# x = 1.0, y = 2.0, w = 1.0
# dLoss/dw = 2 * (w*x - y) * x
# dLoss/dw = 2 * (1.0 * 1.0 - 2.0) * 1.0 = 2 * (-1.0) * 1.0 = -2.0

# Substitute the numerical values into the symbolic derivative to see the correspondence
numerical_dloss_dw = dloss_dw_sym.subs({x_sym: 1.0, y_sym: 2.0, w_sym: 1.0})
print(f"Numerical value of the symbolic derivative (at x=1, y=2, w=1): {numerical_dloss_dw}")
```
